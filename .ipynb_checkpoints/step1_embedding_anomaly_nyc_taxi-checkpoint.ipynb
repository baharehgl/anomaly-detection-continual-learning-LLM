# Step 1: Embedding-Based Anomaly Score on NYC Taxi (NAB)

**Dataset:** NYC Taxi passengers, 30-minute buckets, 2014-07-01 to 2015-01-31 (10,320 rows). Source: Numenta Anomaly Benchmark (NAB). Five known anomalies: NYC Marathon, Thanksgiving, Christmas, New Year's Day, Snow Storm.

**Method:** for each window of the series, extract TimesFM 2.5 hidden states from a middle transformer layer (layer 10 of 20), then score by Mahalanobis distance to the cloud of normal embeddings.

**Why this beats residual+IQR:** residuals miss anomalies the forecaster predicts well (e.g., Christmas — periodic enough to forecast) and false-alarm on rare-but-normal patterns. Embeddings describe what the point *looks like* in the model's learned space.

References: THEMIS (arXiv 2510.03911), TimeRep (arXiv 2509.12650), NAB (Lavin & Ahmad 2015).


## 1. Install and import

```python
# %pip install -q timesfm torch numpy pandas matplotlib scikit-learn kagglehub
```

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt, torch
import timesfm

assert torch.cuda.is_available(), "GPU required for TimesFM 2.5"
DEVICE = "cuda"
SEED = 42
np.random.seed(SEED); torch.manual_seed(SEED)
print("torch:", torch.__version__, "| cuda:", torch.cuda.get_device_name(0))
```

## 2. Load NYC Taxi data + NAB ground-truth labels

Two ways to get the CSV. Either:

**(A) Programmatic via kagglehub** — needs `~/.kaggle/kaggle.json` API token set up.

**(B) Manual** — download `nyc_taxi.csv` from Kaggle and put it next to this notebook.

The five anomaly windows come from the canonical NAB labels file (`numenta/NAB/labels/combined_windows.json`). Every timestamp inside one of these windows is labeled `is_true_anomaly = True`.


```python
# NAB-labeled anomaly windows for nyc_taxi.csv. Source:
# https://github.com/numenta/NAB/blob/master/labels/combined_windows.json
NAB_ANOMALY_WINDOWS = [
    ("2014-10-30 15:30:00", "2014-11-03 22:30:00"),   # NYC Marathon
    ("2014-11-25 12:00:00", "2014-11-29 19:00:00"),   # Thanksgiving
    ("2014-12-23 11:30:00", "2014-12-27 18:30:00"),   # Christmas
    ("2014-12-29 21:30:00", "2015-01-03 04:30:00"),   # New Year's Day
    ("2015-01-24 20:30:00", "2015-01-29 03:30:00"),   # Snow Storm
]

def load_nyc_taxi(csv_path: str | None = None) -> pd.DataFrame:
    # Option A: kagglehub
    if csv_path is None:
        import kagglehub, os
        root = kagglehub.dataset_download("julienjta/nyc-taxi-traffic")
        # The dataset typically contains nyc_taxi.csv at the root
        for fn in os.listdir(root):
            if fn.endswith(".csv"):
                csv_path = os.path.join(root, fn); break
        assert csv_path, f"no csv found in {root}"

    df = pd.read_csv(csv_path)
    # Normalize column names just in case
    df.columns = [c.strip().lower() for c in df.columns]
    df["timestamp"] = pd.to_datetime(df["timestamp"])
    df = df.sort_values("timestamp").reset_index(drop=True)

    # Apply NAB windows as ground truth
    is_anom = np.zeros(len(df), dtype=bool)
    for start, end in NAB_ANOMALY_WINDOWS:
        mask = (df["timestamp"] >= start) & (df["timestamp"] <= end)
        is_anom |= mask.values
    df["is_true_anomaly"] = is_anom
    return df[["timestamp", "value", "is_true_anomaly"]]

df = load_nyc_taxi()                # or: load_nyc_taxi("nyc_taxi.csv")
print(f"rows: {len(df)} | columns: {list(df.columns)}")
print(f"range: {df.timestamp.min()} → {df.timestamp.max()}")
print(f"anomaly samples: {df.is_true_anomaly.sum()} ({df.is_true_anomaly.mean():.1%})")
df.head()
```

```python
# Quick look at the raw series with the NAB windows highlighted
fig, ax = plt.subplots(figsize=(13, 3.5))
ax.plot(df.timestamp, df.value, lw=0.5, color="steelblue")
for start, end in NAB_ANOMALY_WINDOWS:
    ax.axvspan(pd.Timestamp(start), pd.Timestamp(end), color="red", alpha=0.18)
ax.set_title("NYC Taxi passengers (30-min buckets) — red = NAB anomaly windows")
ax.set_ylabel("passengers / 30 min"); plt.tight_layout(); plt.show()
```

## 3. Load TimesFM 2.5 and hook a middle layer

`CTX_LEN = 1024` ≈ 21 days at 30-min granularity — captures 3 full weeks of seasonality, which matters because the NYC taxi series has strong daily (period=48) and weekly (period=336) cycles.

`LAYER_IDX = 10` — middle of the 20-layer stack. Sweep later if you want to tune (try 8 and 12).


```python
CTX_LEN   = 1024       # ~21 days of 30-min buckets
LAYER_IDX = 10         # middle of the 20-layer transformer stack

tfm = timesfm.TimesFm_2p5(
    hparams=timesfm.TimesFmHparams(
        backend=DEVICE,
        per_core_batch_size=16,
        horizon_len=48,
        context_len=CTX_LEN,
    ),
    checkpoint=timesfm.TimesFmCheckpoint(
        huggingface_repo_id="google/timesfm-2.5-200m-pytorch"
    ),
)
tfm.model.eval()
for p in tfm.model.parameters(): p.requires_grad_(False)

# Forward hook on the chosen transformer block
_cache = {}
def _hook(module, inp, out):
    h = out[0] if isinstance(out, tuple) else out
    _cache["h"] = h.detach()

target = tfm.model.stacked_transformer.layers[LAYER_IDX]
handle = target.register_forward_hook(_hook)
print(f"hook on layer {LAYER_IDX}; ctx_len={CTX_LEN}")
```

## 4. Slide windows and embed every position

Stride = 24 (12 hours). Smaller stride = denser scoring but more compute. With ~10,000 rows and stride 24, you get ~400 windows total — runs in a couple of minutes on a single GPU.


```python
STRIDE = 24
BATCH  = 16

def make_windows(values, ctx_len=CTX_LEN, stride=STRIDE):
    xs, ends = [], []
    for end in range(ctx_len, len(values), stride):
        xs.append(values[end - ctx_len:end])
        ends.append(end - 1)
    return np.stack(xs).astype(np.float32), np.array(ends)

@torch.no_grad()
def embed_batch(w):
    x = torch.from_numpy(w).to(DEVICE)
    _ = tfm.model(x)
    return _cache["h"].mean(dim=1).cpu().numpy()      # mean-pool patches -> [B, D]

def embed_all(values):
    w, ends = make_windows(values)
    out = []
    for i in range(0, len(w), BATCH):
        out.append(embed_batch(w[i:i+BATCH]))
    return np.vstack(out), ends

# Use the first 50% as the "normal" reference set.
# (For NYC taxi this is roughly July–October 2014, all pre-Marathon.)
split = len(df) // 2
train_emb, _         = embed_all(df.value.values[:split])
all_emb,   all_ends  = embed_all(df.value.values)
print("train_emb:", train_emb.shape, "| all_emb:", all_emb.shape)
```

## 5. Mahalanobis score

Mean and covariance of the normal cloud, then Mahalanobis distance for every window.


```python
def fit_gaussian(X, ridge=1e-3):
    mu = X.mean(axis=0)
    Xc = X - mu
    cov = (Xc.T @ Xc) / max(len(X) - 1, 1)
    cov += ridge * np.eye(cov.shape[0])
    return mu, np.linalg.inv(cov)

def mahalanobis(X, mu, cov_inv):
    Xc = X - mu
    return np.sqrt(np.einsum("ij,jk,ik->i", Xc, cov_inv, Xc))

mu, cov_inv = fit_gaussian(train_emb)
scores_w = mahalanobis(all_emb, mu, cov_inv)
print(f"score stats: min={scores_w.min():.2f}  median={np.median(scores_w):.2f}  max={scores_w.max():.2f}")
```

## 6. Map per-window scores back to per-timestamp scores

```python
def align_to_series(ends, scores, n_total):
    aligned = np.full(n_total, np.nan)
    aligned[ends] = scores
    last = np.nan
    for i in range(n_total):
        if not np.isnan(aligned[i]): last = aligned[i]
        elif not np.isnan(last): aligned[i] = last
    return aligned

scores_series = align_to_series(all_ends, scores_w, len(df))

fig, ax = plt.subplots(2, 1, figsize=(13, 6), sharex=True)
ax[0].plot(df.timestamp, df.value, lw=0.4, color="steelblue")
for s, e in NAB_ANOMALY_WINDOWS:
    ax[0].axvspan(pd.Timestamp(s), pd.Timestamp(e), color="red", alpha=0.15)
ax[0].set_ylabel("passengers")
ax[1].plot(df.timestamp, scores_series, lw=0.6, color="purple")
for s, e in NAB_ANOMALY_WINDOWS:
    ax[1].axvspan(pd.Timestamp(s), pd.Timestamp(e), color="red", alpha=0.15)
ax[1].set_ylabel("Mahalanobis score"); ax[1].set_xlabel("time")
plt.tight_layout(); plt.show()
```

## 7. Evaluate vs. residual+IQR baseline

Both methods flag the top-X% of points. We pick X = anomaly rate in the data (~10% for this dataset because the NAB windows are wide) so precision/recall are comparable.


```python
def evaluate(flags, truth):
    flags = flags.astype(bool); truth = truth.astype(bool)
    tp = int((flags & truth).sum()); fp = int((flags & ~truth).sum()); fn = int((~flags & truth).sum())
    prec = tp / max(tp+fp, 1); rec = tp / max(tp+fn, 1)
    f1 = 2*prec*rec / max(prec+rec, 1e-9)
    return {"precision": round(prec,3), "recall": round(rec,3), "f1": round(f1,3),
            "tp": tp, "fp": fp, "fn": fn}

# Baseline: residual + IQR (rolling-mean residual, window=48 = one day)
roll = pd.Series(df.value.values).rolling(window=48, center=True, min_periods=12).mean()
residual = df.value.values - roll.values
mask = np.isfinite(residual)
q1, q3 = np.percentile(residual[mask], 25), np.percentile(residual[mask], 75)
iqr_v = q3 - q1
flags_iqr = np.zeros(len(df), dtype=bool)
flags_iqr[mask] = (residual[mask] < q1 - 3*iqr_v) | (residual[mask] > q3 + 3*iqr_v)

# New: embedding Mahalanobis, threshold at the anomaly-rate-matched percentile
finite_s = scores_series[np.isfinite(scores_series)]
anom_rate = df.is_true_anomaly.mean()                # ~0.10 for NYC taxi
pct = 100 * (1 - anom_rate)
thr = np.percentile(finite_s, pct)
flags_emb = np.isfinite(scores_series) & (scores_series >= thr)

print(f"true anomaly rate: {anom_rate:.1%}  | embedding threshold = {pct:.1f}th percentile")
print("residual + IQR :", evaluate(flags_iqr,  df.is_true_anomaly.values))
print("embedding score:", evaluate(flags_emb,  df.is_true_anomaly.values))
```

## 8. Per-event breakdown

Did we catch each of the five named events? This is a more honest evaluation than aggregate precision/recall — NAB scoring also cares whether you detect each event at all, not just how many points you flag.


```python
def event_hits(flags, windows, ts):
    rows = []
    for s, e in windows:
        m = (ts >= pd.Timestamp(s)) & (ts <= pd.Timestamp(e))
        rows.append({"window": f"{s} → {e}",
                     "duration_pts": int(m.sum()),
                     "flagged_pts": int((flags & m).sum()),
                     "hit": bool((flags & m).any())})
    return pd.DataFrame(rows)

print("=== residual + IQR ===")
print(event_hits(flags_iqr, NAB_ANOMALY_WINDOWS, df.timestamp).to_string(index=False))
print("\n=== embedding score ===")
print(event_hits(flags_emb, NAB_ANOMALY_WINDOWS, df.timestamp).to_string(index=False))
```

## 9. What to try next

If the embedding score doesn't clearly beat residual+IQR, sweep in this order:

1. **`LAYER_IDX`**: 8, 12. Re-run from cell 4. (`handle.remove()` first to drop the old hook.)
2. **`CTX_LEN`**: 2048. Captures more weekly structure but slower.
3. **Pooling**: try `max` or `last-token` pooling instead of `mean` in `embed_batch`.
4. **Robust covariance**: `from sklearn.covariance import MinCovDet; mcd = MinCovDet().fit(train_emb); mu, cov_inv = mcd.location_, np.linalg.inv(mcd.covariance_)`. Helps if anomalies leaked into the train slice.
